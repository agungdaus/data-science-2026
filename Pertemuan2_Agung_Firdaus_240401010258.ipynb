````python
#Q1 Berapa jumlah baris dan kolom pada dataset Titanic? Kolom apa saja yang tersedia?
import pandas as pd
import numpy as np

URL = "https://raw.github.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)

# Dimensi dataset
print(f"Baris : {df.shape[0]}")
print(f"Kolom : {df.shape[1]}")
print(f"\nKolom yang tersedia:\n{df.columns.tolist()}")
````

````python
#Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?
# Hitung jumlah dan proporsi missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

result = pd.DataFrame({
    'Missing': missing,
    'Proporsi (%)': missing_pct
})
print(result[result['Missing'] > 0])
````

````python
#Q3 Berapa persen penumpang yang selamat (Survived=1)? Gunakan operasi NumPyatau Pandas.
# Menggunakan NumPy
survived_pct_np = np.mean(df['Survived'].values) * 100

print(f"[NumPy]  Selamat: {survived_pct_np:.2f}%")
print(f"Jumlah selamat : {df['Survived'].sum()} dari {len(df)} penumpang")
````

````python
#Q4 Filter penumpang wanita dari kelas 1 (Pclass=1). Berapa jumlahnya dan berapa rata-rata usianya?
# Filter: Sex == 'female' AND Pclass == 1
wanita_kelas1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]

jumlah = len(wanita_kelas1)
rata_usia = wanita_kelas1['Age'].mean()

print(f"Jumlah wanita kelas 1 : {jumlah}")
print(f"Rata-rata usia        : {rata_usia:.2f} tahun")
````

````python
# Q5 Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?
# Groupby Pclass → mean Survived
survival_rate = df.groupby('Pclass')['Survived'].mean() * 100
survival_rate = survival_rate.round(2)

print(survival_rate)
print(f"\nKelas tertinggi: Kelas {survival_rate.idxmax()} ({survival_rate.max()}%)")
````

````python
#Buat analisis tambahan berdasarkan variabel lain yang Anda anggap menarik.Sertakan interpretasi singkat.
# Analisis: tingkat keselamatan berdasarkan Sex × Pclass
pivot = df.pivot_table(
    values='Survived',
    index='Sex',
    columns='Pclass',
    aggfunc='mean'
) * 100

print(pivot.round(2))

# Bonus: survival berdasarkan embarkasi
embarked_surv = df.groupby('Embarked')['Survived'].mean() * 100
print("\nSurvival per embarkasi:\n", embarked_surv.round(2))

#Analisis silang Sex × Pclass mengungkap dampak kebijakan "women and children first" yang sangat jelas:
#wanita kelas 1 selamat hampir semua (96.81%), sementara pria kelas 2 hanya 15.74% — bahkan lebih rendah dari wanita kelas 3 (50%).
````
